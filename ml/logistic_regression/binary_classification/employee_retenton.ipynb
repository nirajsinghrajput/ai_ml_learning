```python
import pandas as pd
df = pd.read_csv('HR_comma_sep.csv')
df.head(15)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.xlabel('Salaries')
plt.ylabel('retention')
plt.bar(df.salary,df.left)
```

```python
pd.crosstab(df.salary,df.left).plot(kind='bar')
```

```python
plt.xlabel('Department')
plt.ylabel('retention')
plt.bar(df.Department,df.left)
plt.xticks(rotation=45, ha= 'right' )
```

```python
dummies1 = pd.get_dummies(df.salary).astype(int)
dummies1.head()
```

```python
dummies2 = pd.get_dummies(df.Department).astype(int)
dummies2
```

```python
merged = pd.concat([df,dummies1,dummies2],axis='columns')
merged
```

```python
final = merged.drop(['salary','medium','Department','IT'],axis='columns')
final
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = final.drop('left',axis='columns')
x
```

```python
y = df.left
y
```

```python
X_train , X_test , y_train , y_test = train_test_split(x,y,test_size=0.9)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=15000)
```

```python
model.fit(X_train,y_train)
```

```python
X_test
```

```python
model.predict(X_test.head())
```

```python
y_test.head()
```

```python
model.score(X_test,y_test)
```
