```python
import pandas as pd
```

```python
df = pd.read_csv('carprices.csv')
df
```

```python
dummies = pd.get_dummies(df['Car Model']).astype(int)
dummies
```

```python
merged = pd.concat([df,dummies],axis='columns')
merged
```

```python
final = merged.drop(['Car Model','Mercedez Benz C class'],axis='columns')
final
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
x = final.drop('Sell Price($)',axis='columns')
x
```

```python
y = final['Sell Price($)']
y
```

```python
model.fit(x,y)
```

```python
model.predict([[45000,4,0,0]])
```

```python
model.predict([[86000,7,0,1]])
```

```python
model.score(x,y)
```

# Label encoding

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
dfle = df
dfle['Car Model'] = le.fit_transform(dfle['Car Model'])
dfle
```

```python
xle = dfle[['Car Model','Mileage','Age(yrs)']].values
xle
```

```python
yle = dfle['Sell Price($)']
yle
```

```python
model.fit(xle,yle)
```

```python
model.predict([[2,45000,4]])
```

```python
model.predict([[1,86000,7]])
```

```python
model.score(xle,yle)
```

# One hot encoder

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(drop='first', sparse_output=False),[0])
    ],
    remainder='passthrough'
)
xohe = ct.fit_transform(xle)
xohe
```

```python
model.fit(xohe,yle)
```

```python
model.predict([[0,1,45000,4]])
```

```python
model.predict([[1,0,86000,7]])
```

```python
model.score(xohe,yle)
```
